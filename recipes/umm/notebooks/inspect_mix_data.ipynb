```python
from recipes.datasets.mcc.mix_mkii import MixDataModule, collate_all
from transformers import BertTokenizer
import IPython.display as ipd


sample_rate = 24000
pl_datamodule = MixDataModule(
    sample_rate=sample_rate,
    batch_size=sample_rate * 7 * 90,
    shuffle_buffer_size=50,
    max_num_crops=None,
    min_duration=1,
    max_duration=90,
    num_workers=3,
    collate_fn=collate_all,
    tokenizer="bert-base-multilingual-uncased",
)
train_loader = pl_datamodule.train_dataloader()
train_loader = iter(train_loader)
```

```python
batch = next(train_loader)
for token, text, audio in zip(batch["token"], batch["text"], batch["audio"]):
    print(text)
    print(pl_datamodule.tokenizer.decode(token, skip_special_tokens=True, clean_up_tokenization_spaces=True))
    ipd.display(ipd.Audio(audio, rate=sample_rate))
```
